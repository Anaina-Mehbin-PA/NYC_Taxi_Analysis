## **2. About Dataset**

The dataset used in this project focuses on NYC Yellow Taxi trip data for the year 2024. The dataset provides information about taxi trips, including trip distance, fare amount, passenger count, payment type, pickup and drop-off boroughs, and trip duration.

This dataset can be used to analyze taxi usage patterns, payment methods, trip distances, fares, passenger behavior, and travel patterns throughout the day.

## **3. Dataset Features**

    Date: Date on which the taxi trips occurred.

    Hour: Hour of the day when the trips occurred.

    Passenger Count: Number of passengers in the taxi.

    Pickup Borough: Borough where the taxi trip started.

    Drop-off Borough: Borough where the taxi trip ended.

    Payment Type: Method of payment used for the trip.

    Trip Count: Number of taxi trips.

    Trip Distance: Total distance covered by the trips.

    Duration: Total duration of the trips.

    Fare Amount: Total fare amount generated from the trips.

    Tip Amount: Total tip amount received.

    Total Amount: Total amount collected from the trips.

## **6. Feature Extraction**

Feature extraction was performed to derive additional useful information from the existing columns. Two new features, Average Fare and Average Tip, were created to understand the average fare and tip generated per taxi trip.

Average Fare: Represents the average fare amount for each trip.

## Overall Key Insights

* NYC Yellow Taxi usage varies across different hours of the day, showing clear differences in trip activity throughout the day.
* Pickup activity is not evenly distributed across the NYC boroughs, with some boroughs recording more taxi trips than others.
* The analysis shows that payment methods are not equally used, indicating differences in passenger payment preferences.
* Passenger counts show that taxi trips are more common for certain passenger group sizes.
* Trip distance and fare amount show a relationship, indicating that longer taxi trips generally involve higher fare amounts.
* Trip duration and trip distance also show a relationship, helping to understand how travel distance is associated with the time taken for trips.
* The boxplot analysis identifies variation and potential outliers in trip distance and fare amounts.
* Average fare and average tip provide additional information about the amount generated per taxi trip.
* Overall, the analysis provides useful insights into taxi usage, passenger behavior, payment preferences, trip characteristics, and fare patterns in NYC during 2024.
